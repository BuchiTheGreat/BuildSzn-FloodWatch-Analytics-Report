```python
import rasterio
import numpy as np
import pandas as pd

# 1. Path to your downloaded elevation tile
tif_file = "n06_e003_1arc_v3.tif"

print("Reading GeoTIFF and extracting coordinate matrices...")
with rasterio.open(tif_file) as src:
    # Read the elevation values from the first band
    elevation_array = src.read(1)
    
    # Generate grid grids of pixel positions (rows and columns)
    height, width = elevation_array.shape
    cols, rows = np.meshgrid(np.arange(width), np.arange(height))
    
    # Convert pixel indices to real geographic Longitude (x) and Latitude (y)
    xs, ys = rasterio.transform.xy(src.transform, rows, cols)
    
    # Flatten the 2D pixel grids into 1D arrays for the table
    longitudes = np.array(xs).flatten()
    latitudes = np.array(ys).flatten()
    elevations = elevation_array.flatten()

# 2. Build the DataFrame to match your project specifications
print("Building DataFrame...")
df = pd.DataFrame({
    'Latitude': latitudes,
    'Longitude': longitudes,
    'Elevation_Meters': elevations
})

# --- OPTIONAL SPEED BOOST FOR POWER BI ---
# 13 million rows can make Power BI maps sluggish. 
# Uncomment the line below to take every 4th coordinate point instead (reduces data to ~800k rows while keeping the map perfectly accurate).
# df = df.iloc[::4].reset_index(drop=True)
# ------------------------------------------

# 3. Export to CSV
output_csv = "lagos_elevation.csv"
print(f"Exporting data to {output_csv}... (This might take a minute due to file size)")
df.to_csv(output_csv, index=False)

print(f"Done! Successfully generated {len(df):,} rows of elevation data.")
```
